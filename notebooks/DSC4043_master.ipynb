# 1. Data Preparation & Cleaning — *Aliff*

The full preparation pipeline is in `src/01_data_prep.py`. Summary of what it does:

1. Loads both raw CSVs (semicolon-delimited)
2. Inner-merges on the 13 student-identity attributes → **382 students**
3. Outer-merges to quantify and document the missing data
4. Removes duplicates, strips whitespace, fixes dtypes
5. Winsorises `absences` outliers by the IQR rule
6. Derives `G3_avg`, `G3_diff`, `band_mat`, `band_por`, `pass_mat`, `pass_por`
7. Exports `student_clean.csv`

```python
# Evidence of the merge and the cleaned result
print("Analysis dataset:", df.shape)
print("\nMerge keys (13, identical in both subjects):")
print(["school","sex","age","address","famsize","Pstatus","Medu",
       "Fedu","Mjob","Fjob","reason","nursery","internet"])

print("\nSubject-suffixed columns:")
print(sorted({c.rsplit("_",1)[0] for c in df.columns if c.endswith(("_mat","_por"))}))

print("\nFirst five rows, key columns:")
display(df[["school","sex","age","G3_mat","G3_por","G3_avg","band_mat"]].head())
```
